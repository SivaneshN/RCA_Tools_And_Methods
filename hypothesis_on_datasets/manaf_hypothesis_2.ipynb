```python
# Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

dataset_path = "../data/merged_dataset.csv"

df = pd.read_csv(dataset_path)

df.info()
```

### Gini Index Differences (Mean Comparison)

Question: Do countries with high inequality have higher homicide levels?

### Creating two groups

Split by median Gini value:

```python
median_gini = df['gini_index'].median()
high = df[df['gini_index'] > median_gini]['homicide_data']
low  = df[df['gini_index'] <= median_gini]['homicide_data']
```

### TWO-TAILED T-TEST (High vs Low Inequality)
Hypotheses

H₀: μ₁ = μ₂ (no difference)

H₁: μ₁ ≠ μ₂ (difference exists)

```python
from scipy.stats import ttest_ind

t_stat, p_val = ttest_ind(high, low, equal_var=False)
print("t-stat:", t_stat)
print("two-tailed p:", p_val)
```

### ONE-TAILED T-TEST

- Test if high-inequality → higher homicide:

```python
p_one = p_val / 2 if t_stat > 0 else 1 - (p_val / 2)
print("One-tailed p:", p_one)
```

### Decision rule

p_one < 0.05 → High inequality significantly increases homicide.

### Z-TEST (Large sample)

- Your dataset has > 1200 rows → Z-test is acceptable.

```python
import numpy as np
mean_high = np.mean(high)
mean_low  = np.mean(low)
var_high  = np.var(high, ddof=1)
var_low   = np.var(low, ddof=1)

z = (mean_high - mean_low) / np.sqrt(var_high/len(high) + var_low/len(low))
print("Z-statistic:", z)
```

Compare with critical Z = ±1.96.
