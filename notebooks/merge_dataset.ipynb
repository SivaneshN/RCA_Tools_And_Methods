## Intentional Death Dataset
- Combine Suicide and Homicide Datasets to make 


```python
# Imports
# Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

suicide_dataset_path = "../data/suicide_data_clean.csv"
homicide_dataset_path = "../data/homicide_data_clean.csv"
dataset_maxrows = 200

suicide_df = pd.read_csv(suicide_dataset_path)
homicide_df = pd.read_csv(homicide_dataset_path)


# Merging on multiple columns ('country_code' and 'year')
intentional_death_df = pd.merge(suicide_df, homicide_df, left_on=['country_code', 'year'], right_on=['country_code', 'year'], how='inner')
intentional_death_df = intentional_death_df.drop(['world_regions','country_name_y','zscore_y','zscore_x'], axis=1)

intentional_death_df['intentional_death'] = intentional_death_df['suicide_data'] + intentional_death_df['homicide_data']  # assigned to a column
intentional_death_df = intentional_death_df.rename(columns={'country_name_x': 'country_name'})

intentional_death_df
```

## Missing Data: Identify → Decide → Apply
- Find empty records

```python
# Identify missing data
missing_counts = intentional_death_df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing_counts / len(intentional_death_df) * 100).round(2)
missing_report = pd.DataFrame({"missing_count": missing_counts, "missing_pct": missing_pct})

missing_report
```

```python
# show the missing rows

bool_series = pd.isnull(intentional_death_df["country_code"])
missing_country_code_data = intentional_death_df[bool_series]

missing_country_code_data
```

```python
unique_country_code = intentional_death_df['country_code'].unique()

#print(unique_countries)
unique_country_code
```

## Final Merged Dataset
- Merging Intentional Dataset and Economic Gini Index 

```python

gini_index_dataset_path = "../data/gini_index_data_clean.csv"
dataset_maxrows = 200

gini_index_df = pd.read_csv(gini_index_dataset_path)


# Merging on multiple columns ('country_code' and 'year')
merged_df = pd.merge(intentional_death_df, gini_index_df, left_on=['country_code', 'year'], right_on=['country_code', 'year'], how='inner')

merged_df = merged_df.drop(['world_regions','country_name_y','zscore'], axis=1)
merged_df = merged_df.rename(columns={'country_name_x': 'country_name'})

merged_df
```
