```python
# Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

dataset_path = "../data/merged_dataset.csv"

df = pd.read_csv(dataset_path)

df.info()
```

### Chi-Square Test (Categorical Relationship)
- Convert Gini into categories

```python
df['gini_cat'] = pd.qcut(df['gini_index'], 3, labels=['low','medium','high'])
df['hom_cat'] = pd.qcut(df['homicide_data'], 3, labels=['low','medium','high'])
df['suicide_cat'] = pd.qcut(df['suicide_data'], 3, labels=['low','medium','high'])
```

### Hypotheses

- H₀: Gini category is independent of homicide category
- H₁: They are associated

### Chi-Square Test of Independence

```python
import pandas as pd
from scipy.stats import chi2_contingency

ct = pd.crosstab(df['gini_cat'], df['hom_cat'])
ct_2 = pd.crosstab(df['gini_cat'], df['suicide_cat'])
chi2, homicide_p, dof, expected = chi2_contingency(ct)
chi2, suicide_p, dof, expected = chi2_contingency(ct_2)


print("Chi-square for homicide:", chi2)
print("p-value for homicide:", homicide_p)


print("Chi-square for suicide:", chi2)
print("p-value for suicide:", suicide_p)
```

### Interpretation

p < 0.05 → Reject H₀ → Economic inequality is associated with homicide/suicide levels
