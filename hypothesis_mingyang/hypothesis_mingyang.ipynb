## Hypothesis Multivariate Hypothesis

```python
# Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt



dataset_path = "../data/merged_dataset.csv"

df = pd.read_csv(dataset_path)

#df.head()


filtered_df = df[(df['year'].isin([2000, 2005])) | (df['year'].isin([2010, 2015]))]
filtered_df.head()
```

### Hypotheses

- H₀: The Gini index does not predict suicide or homicide rates.
This means that economic inequality has no statistically significant effect on intentional death outcomes (suicide and homicide). Any observed differences in suicide or homicide rates are assumed to be due to random variation rather than inequality.
- H₁: The Gini index significantly predicts suicide or homicide rates.
This means that economic inequality has a statistically significant effect on intentional deaths. As inequality increases or decreases, suicide and homicide rates are expected to change in a systematic way.


##### Rationale — Why is this hypothesis relevant?

This hypothesis is directly relevant to your research question—“How does a country’s economic inequality (Gini index) affect intentional death rates?”—because:

- Economic inequality is a key structural factor that influences social stress, access to resources, and social cohesion.

- Suicide and homicide are known to be affected by economic, psychological, and social pressures, all of which are linked to inequality.

- Countries with higher inequality often experience greater psychosocial stress, social fragmentation, and reduced access to support systems, which can elevate the rates of intentional deaths.

- Testing this hypothesis allows you to determine whether inequality is a significant predictor, statistically and substantively, using large-scale data.

- If the Gini index is significant (as your regression results show with p < 0.001), it suggests that policy efforts targeting socioeconomic inequality may help reduce suicide and homicide rates.

The hypothesis is important because it tests whether inequality—an underlying socioeconomic factor—plays a measurable role in shaping rates of intentional harm. This aligns directly with your goal of understanding how inequality influences national-level health and safety outcomes.

### MANOVA 

```python
from statsmodels.multivariate.manova import MANOVA

# MANOVA: suicide & homicide predicted by gini
maov = MANOVA.from_formula("suicide_data + homicide_data ~ gini_index", data=filtered_df)
print(maov.mv_test())
```

p < 0.001

This means the Gini index significantly predicts the combined outcomes (suicide + homicide).

1. Wilks’ Lambda = 0.7565 (p < .001)

Wilks’ Lambda measures how much variance in the dependent variables is not explained by the predictor.
- Values closer to 0 → strong effect
- Values closer to 1 → weak effect
- Your value: 0.7565 → moderate effect
- p < .001 → statistically significant

👉 Interpretation:
Gini index explains a significant portion of the variance in suicide & homicide rates.

2. Pillai’s Trace = 0.2435 (p < .001)

This measures the proportion of explained variance:

👉 24.35% of the variance in the combined outcomes (suicide + homicide) is explained by inequality.

This matches your earlier regression that showed ~15–16% but expands to the multivariate case.

3. Hotelling-Lawley & Roy’s Root

These also confirm the same result:

- Strong effect
- Highly significant
- Consistent across all multivariate tests

### Final Conclusion:

Economic inequality (Gini index) has a statistically significant multivariate effect on intentional death rates (suicide & homicide) across 220 observations. Higher inequality is associated with higher levels of suicide and homicide when examined together.
